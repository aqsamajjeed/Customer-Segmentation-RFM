```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Read the Excel file
df = pd.read_excel("C:\\Users\\Fame\\Downloads\\Online Retail (1).xlsx")

# Preview data
df.head()
```

```python
df.dropna(subset=['CustomerID'],inplace=True)

#keeping only positive values
df=df[df['Quantity']>0]

#creating new column for total price
df['Total Price']=df['Quantity']*df['UnitPrice']
df['InvoiceNo'] = df['InvoiceNo'].astype(str).str.strip()
df.head()
```

```python
#define latest date for recency calculation
latest_date=df['InvoiceDate'].max() + pd.Timedelta(days=1)

rfm= df.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (latest_date - x.max()).days,     #recency
    'InvoiceNo': 'count',                                      #frequency
    'Total Price': 'sum'                                       #monetary
})

rfm.rename(columns={
    'InvoiceDate': 'Recency',
    'InvoiceNo': 'Frequency',
    'Total Price': 'Monetary'
}, inplace=True)


rfm.head()
```

```python
# Function to assign scores dynamically
def assign_scores(series, reverse=False):
    bins = pd.qcut(series, 4, labels=False, duplicates='drop')
    if reverse:
        bins = (bins.max() - bins) + 1   # higher score for lower values
    else:
        bins = bins + 1
    return bins

# Assign scores
rfm['R_Score'] = assign_scores(rfm['Recency'], reverse=True)
rfm['F_Score'] = assign_scores(rfm['Frequency'])
rfm['M_Score'] = assign_scores(rfm['Monetary'])

# Total RFM Score
rfm['RFM_Score'] = rfm['R_Score'] + rfm['F_Score'] + rfm['M_Score']

rfm.head()
```

```python
# Segmentation function
def segment(score):
    if score >= 9:
        return 'Best Customers'  # Discounts, VIP perks
    elif score >= 8:
        return 'Loyal Customers' # Rewards, exclusive offers
    elif score >= 5:
        return 'Potential Loyalist' # Nurture with personalized emails
    else:
        return 'Needs Attention'  # Re-engagement campaigns

# Apply segmentation
rfm['Segment'] = rfm['RFM_Score'].apply(segment)

rfm.head()
```

```python
# Improved visualization with style changes
plt.figure(figsize=(10, 6))

# Choose a clean style
sns.set_style("whitegrid")

# Use a pastel palette
sns.countplot(
    data=rfm,
    x='Segment',
    order=rfm['Segment'].value_counts().index,
    palette='pastel',
    edgecolor='black'
)


# Titles and labels
plt.title('Customer Segments Based on RFM Analysis', fontsize=16, fontweight='bold')
plt.xlabel('Customer Segment', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)

# Rotate x labels for readability
plt.xticks(rotation=20)

# Remove top/right borders
sns.despine()

plt.show()
```

```python
# --- Visualization: Heatmap (Bonus) ---
plt.figure(figsize=(8,6))
heatmap_data = rfm.groupby(['R_Score','F_Score']).size().unstack(fill_value=0)
sns.heatmap(heatmap_data, cmap="Blues", annot=True, fmt="d")
plt.title("R vs F Heatmap (Customer Distribution)", fontsize=14, fontweight="bold")
plt.show()
```
